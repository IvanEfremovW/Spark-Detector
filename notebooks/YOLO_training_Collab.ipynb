```python
from ultralytics import YOLO
```

```python
model = YOLO("models/yolo26s.pt")
dataset_yaml = "/content/data/data.yaml"
```

```python
results = model.train(
    data=dataset_yaml,

    epochs=150,
    imgsz=736,
    batch=8,
    patience=20,

    device=0,
    workers=16,

    # Augmentations
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,

    mosaic=1.0,

    fliplr=0.5,
    degrees=10.0,
    translate=0.1,
    scale=0.5,
)
```
